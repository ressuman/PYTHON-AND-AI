```python
import pandas as pd

# Original data
grades = pd.DataFrame(
    {
        "Student": ["Ama", "Kojo"],
        "Math": [85, 78],
        "Science": [90, 88],
        "English": [75, 80],
    }
)

print("Original Wide Format:")
print(grades)


# Step 1: Wide -> Long using melt
grades_long = pd.melt(
    grades,
    id_vars="Student",  # keep Student as identifier
    var_name="Subject",  # new column for former headers
    value_name="Score",  # new column for values
)

print("\nLong Format:")
print(grades_long)


# Step 2: Long -> Wide using pivot
grades_wide = grades_long.pivot(
    index="Student", columns="Subject", values="Score"
).reset_index()

# Optional: remove column name display
grades_wide.columns.name = None

print("\nBack to Wide Format:")
print(grades_wide)
```
