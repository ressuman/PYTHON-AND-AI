```python
# Melt into long format
df_long = df.melt(
    id_vars=["employee_id", "name"],  # identifiers
    value_vars=["department", "age", "salary", "join_date", "rating"],  # features
    var_name="feature",
    value_name="value",
)

print(df_long.head(15))
```

```python
import pandas as pd
import numpy as np

# --- Create messy dataset ---
df = pd.DataFrame(
    {
        "Customer_ID": [
            1001,
            "ID1002",
            None,
            1004,
            -1005,
            1006,
            "one007",
            1008,
            1009,
            1010,
            "??",
            1012,
            1013,
            None,
            1015,
            "cust16",
            1017,
            1018,
            1018,
            "XYZ",
        ],
        "Name": [
            "John",
            "",
            None,
            "Mary ",
            "KOFI",
            "esi",
            "123",
            "Akos@",
            "Nana",
            "Kwame",
            "Abena",
            "Kojo2",
            "Linda##",
            "Michael",
            "Sarah",
            "Daniel",
            " ",
            "Paul",
            "Paul",
            "Unknown123",
        ],
        "Age": [
            25,
            "thirty",
            None,
            -5,
            200,
            "26yrs",
            "??",
            27,
            "old",
            24,
            0,
            "twenty one",
            26,
            None,
            999,
            23,
            "NaN",
            28,
            28,
            "unknown",
        ],
        "Gender": [
            "M",
            "F",
            "Male",
            "female",
            "m",
            None,
            "Unknown",
            "X",
            "Boy",
            "Girl",
            "",
            "FEMALE",
            "Male123",
            "M",
            "f",
            "N/A",
            "Female",
            "M",
            "M",
            "??",
        ],
        "City": [
            "Accra",
            "Kumasi ",
            None,
            "123",
            "ACCRA",
            "accra",
            "Ksi",
            "Tamale",
            "Tema",
            "Tema ",
            "Unknown",
            "",
            "Kum@si",
            "TEMA",
            "Kumasi",
            "Acc",
            "Village",
            "NULL",
            "Tema",
            "??",
        ],
        "Salary": [
            5000,
            "7000gh",
            None,
            -100,
            9999999,
            "five thousand",
            6200,
            0,
            "???",
            5300,
            "NaN",
            4700,
            999999999,
            None,
            "6100",
            "salary",
            6400,
            5000,
            5000,
            "",
        ],
        "Loan_Status": [
            "Approved",
            "Rejected",
            None,
            "Approved",
            "Yes",
            "No",
            None,
            "Pending",
            "Rejected",
            "Approved",
            None,
            "Rejected",
            "Approved",
            "Reject",
            None,
            "Approved",
            "Approved",
            None,
            "Rejected",
            "Approved",
        ],
        "Credit_Score": [
            700,
            None,
            -50,
            1200,
            "bad",
            710,
            None,
            640,
            660,
            "excellent",
            715,
            0,
            680,
            None,
            9999,
            "low",
            710,
            700,
            700,
            "???",
        ],
    }
)

# --- Normalize column names ---
df.columns = df.columns.str.strip().str.lower()

# --- Copy for cleaning ---
df_clean = df.copy()

# --- customer_id ---
df_clean["customer_id"] = pd.to_numeric(
    df_clean["customer_id"].astype(str).str.extract(r"(\d+)")[0], errors="coerce"
)

# --- name ---
df_clean["name"] = (
    df_clean["name"]
    .astype(str)
    .str.replace(r"[^a-zA-Z\s]", "", regex=True)
    .str.strip()
    .str.title()
    .replace({"": np.nan})
)

# --- age ---
df_clean["age"] = pd.to_numeric(
    df_clean["age"].astype(str).str.extract(r"(\d+)")[0], errors="coerce"
)
df_clean.loc[(df_clean["age"] < 0) | (df_clean["age"] > 120), "age"] = np.nan

# --- gender ---
df_clean["gender"] = (
    df_clean["gender"]
    .astype(str)
    .str.lower()
    .str.strip()
    .replace(
        {
            "m": "Male",
            "male": "Male",
            "f": "Female",
            "female": "Female",
            "boy": "Male",
            "girl": "Female",
        }
    )
)
df_clean.loc[~df_clean["gender"].isin(["Male", "Female"]), "gender"] = "Other"

# --- city ---
df_clean["city"] = (
    df_clean["city"]
    .astype(str)
    .str.strip()
    .str.title()
    .replace({"": np.nan, "123": np.nan, "??": np.nan, "Null": np.nan})
)

# --- salary ---
df_clean["salary"] = pd.to_numeric(
    df_clean["salary"].astype(str).str.replace(r"[^\d]", "", regex=True),
    errors="coerce",
)
df_clean.loc[(df_clean["salary"] < 100) | (df_clean["salary"] > 100000), "salary"] = (
    np.nan
)

# Numeric columns → mean or median
df_clean["age"] = df_clean["age"].fillna(df_clean["age"].median())
df_clean["salary"] = df_clean["salary"].fillna(df_clean["salary"].mean())

# Categorical columns → mode (most frequent value)
df_clean["gender"] = df_clean["gender"].fillna(df_clean["gender"].mode()[0])
df_clean["city"] = df_clean["city"].fillna(df_clean["city"].mode()[0])

# Binary/Yes-No columns → mode
df_clean["loan_status"] = df_clean["loan_status"].fillna(
    df_clean["loan_status"].mode()[0]
)

# Drop if too many NaNs (example: if >70% missing)
df_clean = df_clean.drop(columns=["credit_score"])  # only if mostly empty


print("✅ Cleaned Wide Format:")
print(df_clean.head())

# --- Long format ---
df_long = df_clean.melt(
    id_vars=["customer_id", "name"],  # identifiers
    var_name="feature",
    value_name="value",
)

print("\n🤖 Long Format:")
print(df_long.head(10))

# --- Pivot back to wide ---
df_pivot = df_long.pivot_table(
    index=["customer_id", "name"], columns="feature", values="value", aggfunc="first"
).reset_index()

print("\n🔄 Pivoted Wide Format:")
print(df_pivot.head())
```
