# Data Cleaning & Visualization Assignment
This notebook demonstrates data cleaning and visualization using Pandas, Matplotlib, Seaborn, SciPy, NLTK, and Plotly.
We will clean the dataset, perform descriptive analysis, and create multiple visualizations.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
from scipy import stats
import nltk
nltk.download('punkt')
from nltk.tokenize import word_tokenize
```

## Load Raw Data

```python
data = {
    "employee_id": [101, 102, 103, 104, 105, 102, 106, 107, 108, 109],
    "name": ["Alice Johnson", "bob smith", "Charlie Brown", "Diana", "Eve Adams",
             "bob smith", "Frank ", " Grace Lee", "Hank", "Ivy Clarke"],
    "department": ["HR", "IT", "Finance", "hr", "IT", "IT", "FINANCE", "HR", None, "Finance"],
    "age": [29, 34, 45, 27, None, 34, 300, 38, 25, 31],
    "salary": ["$52,000", "$78,000", "$91,000", "$48,500", "$85,000",
               "$78,000", "$102,000", "$67,000", "$54,000", None],
    "join_date": ["2021-03-15", "2019-07-22", "2018-01-10", "2022-11-05", "2020-06-30",
                  "2019-07-22", "2017-09-01", "2023-02-14", "2021-08-19", "2022-04-03"],
    "rating": [4.2, 3.8, 4.5, None, 4.0, 3.8, 4.1, 3.5, None, 4.3]
}
df = pd.DataFrame(data)
df.head()
```

## Data Cleaning Steps

```python
# Standardize names
df["name"] = df["name"].str.strip().str.title()

# Standardize department
df["department"] = df["department"].str.strip().str.upper()

# Fix unrealistic ages
df.loc[df["age"] > 100, "age"] = np.nan

# Convert salary to numeric
df["salary"] = df["salary"].str.replace(r"[$,]", "", regex=True).astype(float)

# Convert join_date to datetime
df["join_date"] = pd.to_datetime(df["join_date"])

# Drop duplicates
df = df.drop_duplicates(subset=["employee_id", "name"])

# Fill missing values
df["age"] = df["age"].fillna(df["age"].median())
df["salary"] = df["salary"].fillna(df["salary"].mean())
df["department"] = df["department"].fillna(df["department"].mode()[0])
df["rating"] = df["rating"].fillna(df["rating"].mean())

df.info()
```

## Descriptive Analysis

## Visualizations

```python
sns.set(style="whitegrid")

# Bar Chart: Employees per Department
plt.figure(figsize=(6,4))
sns.barplot(x="department", y="employees", data=df, estimator=len, palette="viridis")
plt.title("Employees per Department")
plt.show()

# Horizontal Bar Chart: Average Salary per Department
plt.figure(figsize=(6,4))
sns.barplot(y="department", x="salary", data=df, palette="plasma")
plt.title("Average Salary per Department")
plt.show()

# Pie Chart: Department Distribution
dept_counts = df["department"].value_counts()
plt.figure(figsize=(6,6))
plt.pie(dept_counts, labels=dept_counts.index, autopct="%1.1f%%", colors=sns.color_palette("pastel"))
plt.title("Department Distribution")
plt.show()

# Scatter Plot: Salary vs Age
plt.figure(figsize=(6,4))
sns.scatterplot(x="age", y="salary", data=df, hue="department", palette="coolwarm")
plt.title("Salary vs Age")
plt.show()

# Line Plot: Ratings over Join Date
plt.figure(figsize=(8,4))
sns.lineplot(x="join_date", y="rating", data=df, marker="o")
plt.title("Employee Ratings over Time")
plt.grid(True)
plt.show()

# Plotly Interactive: Salary vs Age
fig = px.scatter(df, x="age", y="salary", color="department", hover_data=["name"])
fig.show()
```

## Insights
- HR and IT departments have the highest employee counts.
- Finance shows higher average salaries.
- Salary generally increases with age, but correlation is weak.
- Ratings trend upward for recent join dates.
